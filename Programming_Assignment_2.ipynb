**PROBLEM A**

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
sum = float(np.sum(X))
sum
```

```python
mean = sum / X.size
mean
```

```python
sd = float(np.std(X))
sd
```

```python
X_normalized = (X - mean) / sd
X_normalized
```

```python
float(np.mean(X_normalized))
```

```python
float(np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

```python
y = np.arange(1,37)
```
